```python
import pandas as pd
import json

df = pd.read_csv("basic_stats.csv")

def get_minutes_played(stats):
    try:
        return json.loads(stats).get("minutesPlayed", "")
    except (json.JSONDecodeError, TypeError, AttributeError):
        return ""

minutes = df["stats"].apply(get_minutes_played)
# Remove players who did not play or did not dress
excluded_statuses = ["Did Not Play", "Did Not Dress"]

filtered_df = df[~minutes.str.strip().isin(excluded_statuses)].copy()

filtered_df.to_csv("filtered_stats.csv", index=False)

print(f"Original rows: {len(df)}")
print(f"Filtered rows: {len(filtered_df)}")
print(f"Rows removed: {len(df) - len(filtered_df)}")
```

```python
# Convert JSON strings into Python dictionaries
stats_dicts = filtered_df["stats"].apply(
    lambda x: json.loads(x) if isinstance(x, str) else {}
)

# Expand dictionaries into separate columns
stats_columns = pd.json_normalize(stats_dicts)

# Remove the original stats column and combine the expanded columns
filtered_df = filtered_df.drop(columns=["stats"]).reset_index(drop=True)
stats_columns = stats_columns.reset_index(drop=True)

filtered_df = pd.concat([filtered_df, stats_columns], axis=1)
```
